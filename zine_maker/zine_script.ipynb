```python
from PIL import Image, ImageDraw, ImageFont
from datetime import datetime
import math, os

dpi_dict = {
    "72": (595, 842),
    "96": (794, 1123),
    "150": (1240, 1754),
    "300": (2480, 3508)
}

#A4 dimensions at 300dpi
a4_width = dpi_dict["150"][0]
a4_height = dpi_dict["150"][1]

page_width = math.floor(a4_width/2)
page_height = math.floor(a4_height / 4)

now_prefix = str(datetime.now().strftime('%Y-%m-%d-%H-%M-%S'))

os.mkdir("./" + now_prefix)

colors = [(0, 255, 255), (255, 0, 255), (255, 255, 0)]

for i in range(1, 9):
    page = Image.new('RGB',
                    (page_height, page_width),   
                    colors[i%3])  # White
    draw = ImageDraw.Draw(page)

    # use a bitmap font
    font = ImageFont.truetype("./DEVO.ttf", 40)

    draw.text((math.floor(page_height/2)-50, math.floor(page_width/2)-10), "hello", font=font, fill=(0, 0, 0))
    draw.text((page_height-30, page_width-40), str(i), font=font, fill=(0, 0, 0))
    page.save(now_prefix +"/page_" + str(i) + ".png", 'png', quality=100)



# get the middle of the a4im




```

```python
# then paste them together in the right format (the top row should be upside down)

#    Pg5    |   Pg4     |   Pg3     |   Pg2     |
#           |           |           |           |
#           |           |           |           |
#------------------------------------------------
#           |           |           |           |
#           |           |           |           |
#    Pg6    |   Pg 7    |   Pg8     |   Pg1     |

# add the images to an array in order then loop over

# Image.paste(im: Image | str | float | tuple[float, ...], box: Image | tuple[int, int, int, int] | tuple[int, int] | None = None, 
# mask: Image | None = None) → None[source]
# Pastes another image into this image. The box argument is either a 2-tuple giving the upper left corner, 
# a 4-tuple defining the left, upper, right, and lower pixel coordinate, or None (same as (0, 0)). 
# See Coordinate System. If a 4-tuple is given, the size of the pasted image must match the size of the region.

folder_name = "2025-03-13-11-02-04"

a4im = Image.new('RGB',
                 (a4_width, a4_height), 
                 (255, 255, 255))  # White

# page 1 (front cover) gets pasted top right corner
page_1 = Image.open("./"+ folder_name +"/page_1.png")
rotated_page_1 = page_1.transpose(Image.Transpose.ROTATE_90)
a4im.paste(rotated_page_1, (math.floor(a4_width/2), 0))  # top-right corner
# page 2 goes in top left corner
page_2 = Image.open("./"+ folder_name +"/page_2.png")
rotated_page_2 = page_2.transpose(Image.Transpose.ROTATE_270)
a4im.paste(rotated_page_2, (0, 0))  # top-left corner

# page 8 (back cover) below page 1 (front cover)
page_8 = Image.open("./"+ folder_name +"/page_8.png")
rotated_page_8 = page_8.transpose(Image.Transpose.ROTATE_90)
a4im.paste(rotated_page_8, (page_width, page_height))  
# page 3 below page 2
page_3 = Image.open("./"+ folder_name +"/page_3.png")
rotated_page_3 = page_3.transpose(Image.Transpose.ROTATE_270)
a4im.paste(rotated_page_3, (0, page_height))  

# page 7 below page 8
page_7 = Image.open("./"+ folder_name +"/page_7.png")
rotated_page_7 = page_7.transpose(Image.Transpose.ROTATE_90)
a4im.paste(rotated_page_7, (page_width, 2*page_height)) 
# page 4 below page 3
page_4 = Image.open("./"+ folder_name +"/page_4.png")
rotated_page_4 = page_4.transpose(Image.Transpose.ROTATE_270)
a4im.paste(rotated_page_4, (0, 2*page_height)) 

# page 6 below page 7
page_6 = Image.open("./"+ folder_name +"/page_6.png")
rotated_page_6 = page_6.transpose(Image.Transpose.ROTATE_90)
a4im.paste(rotated_page_6, (page_width, 3*page_height)) 
# page 5 below page 4
page_5 = Image.open("./"+ folder_name +"/page_5.png")
rotated_page_5 = page_5.transpose(Image.Transpose.ROTATE_270)
a4im.paste(rotated_page_5, (0, 3*page_height)) 

a4im.save("new_zine.pdf", 'PDF', quality=100)
```
