```python
from PIL import Image, ImageDraw, ImageFont
import math 

#A4 dimensions at 300dpi
a4_width = 2480
a4_height = 3508

page_width = math.floor(a4_width/2)
page_height = math.floor(a4_height / 4)

#im = Image.open(my_image_file)
for i in range(1, 9):
    page = Image.new('RGB',
                    (page_height, page_width),   
                    (255, 255, 255))  # White
    draw = ImageDraw.Draw(page)

    # use a bitmap font
    font = ImageFont.truetype("./DEVO.ttf", 40)

    draw.text((10, 10), str(i), font=font, fill=(0, 0, 0))
    page.save("page_" + str(i) + ".png", 'png', quality=100)

# then paste them together in the right format (the top row should be upside down)

#    Pg4    |   Pg 2    |   Pg3     |   Pg1     |
#           |           |           |           |
#           |           |           |           |
#------------------------------------------------
#           |           |           |           |
#           |           |           |           |
#    Pg5    |   Pg 6    |   BkC     |   FrC     |

# add the images to an array in order then loop over

# Image.paste(im: Image | str | float | tuple[float, ...], box: Image | tuple[int, int, int, int] | tuple[int, int] | None = None, 
# mask: Image | None = None) → None[source]
# Pastes another image into this image. The box argument is either a 2-tuple giving the upper left corner, 
# a 4-tuple defining the left, upper, right, and lower pixel coordinate, or None (same as (0, 0)). 
# See Coordinate System. If a 4-tuple is given, the size of the pasted image must match the size of the region.

# 

#im = Image.open(my_image_file)
a4im = Image.new('RGB',
                 (a4_width, a4_height),   # A4 at 72dpi
                 (255, 255, 255))  # White
#a4im.paste(im, im.getbbox())  # Not centered, top-left corner
a4im.save("new_zine.pdf", 'PDF', quality=100)
```
